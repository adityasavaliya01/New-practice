1.3
Apply the four morphological operations

```python
kernel = np.ones((5, 5), np.uint8)

erosion = cv2.erode(binary, kernel)
dilation = cv2.dilate(binary, kernel)
opening = cv2.morphologyEx(
    binary,
    cv2.MORPH_OPEN,
    kernel
)
closing = cv2.morphologyEx(
    binary,
    cv2.MORPH_CLOSE,
    kernel
)

plt.figure(figsize=(20, 5))

plt.subplot(1, 5, 1)
plt.imshow(binary, cmap="gray")
plt.title("Original Binary")
plt.axis("off")

plt.subplot(1, 5, 2)
plt.imshow(erosion, cmap="gray")
plt.title("Erosion")
plt.axis("off")

plt.subplot(1, 5, 3)
plt.imshow(dilation, cmap="gray")
plt.title("Dilation")
plt.axis("off")

plt.subplot(1, 5, 4)
plt.imshow(opening, cmap="gray")
plt.title("Opening")
plt.axis("off")

plt.subplot(1, 5, 5)
plt.imshow(closing, cmap="gray")
plt.title("Closing")
plt.axis("off")

plt.tight_layout()
plt.show()
```

1.5
WHY morphology works — Markdown

## WHY Morphology Works

Morphological operations modify the shape and structure of objects in a binary image using a structuring element (kernel).

### Erosion

Erosion acts like a local minimum operation. It shrinks white regions and removes small bright objects or noise. It is useful when small unwanted foreground regions need to be removed.

### Dilation

Dilation acts like a local maximum operation. It expands white regions, fills small gaps and can connect nearby foreground regions.

### Opening

Opening = Erosion → Dilation.

Opening is useful for removing small bright noise while preserving the overall shape and size of larger objects.

### Closing

Closing = Dilation → Erosion.

Closing is useful for filling small holes and gaps inside foreground objects without significantly changing the overall object structure.

### Real-World Computer Vision Example

Morphological operations can be applied before a downstream object detector or segmentation algorithm. For example, in a surveillance or industrial inspection system, an image mask may contain small isolated noise or holes. Opening can remove small unwanted noise before detection, while closing can fill small gaps in an object mask before passing the cleaned mask to the downstream detector.

WHY alpha/beta and histograms matter — Markdown

## WHY Alpha/Beta and Histograms Matter

Brightness and contrast can be adjusted using the formula:

**new_pixel = alpha × pixel + beta**

Here, **alpha** controls the contrast gain, while **beta** controls the brightness offset.

### Alpha

Alpha stretches or compresses the pixel-value distribution around its mean.

- alpha > 1 → higher contrast
- 0 < alpha < 1 → lower contrast

For example:

- alpha = 1.5 increases contrast.
- alpha = 0.7 decreases contrast.

### Beta

Beta shifts the complete histogram to the left or right.

- beta > 0 → brighter image
- beta < 0 → darker image

For example:

- beta = 30 makes the image brighter.
- beta = -30 makes the image darker.

### Image Histogram

A histogram shows the distribution of pixel intensities from 0 to 255.

A narrow histogram generally indicates low contrast because most pixel values are concentrated within a small intensity range.

If the histogram is clipped near 0, the image may contain under-exposed regions. If it is clipped near 255, the image may contain over-exposed regions.

Histogram analysis is useful before applying face or object detection because poor brightness or contrast can make visual features harder to detect. Therefore, histogram analysis can help determine whether image preprocessing is required before sending an image to a downstream detector.

3.2
Detect on static images

```python
def draw_yunet_detections(image, faces):

    output = image.copy()

    if faces is None:
        return output

    for face in faces:

        x, y, w, h = face[:4].astype(int)
        score = face[14]
        cv2.rectangle(
            output,
            (x, y),
            (x + w, y + h),
            (0, 255, 0),
            2
        )

        landmarks = face[4:14].reshape(5, 2).astype(int)

        for px, py in landmarks:
            cv2.circle(
                output,
                (px, py),
                3,
                (0, 0, 255),
                -1
            )

        cv2.putText(
            output,
            f"Face: {score:.2f}",
            (x, max(y - 10, 20)),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.6,
            (255, 0, 0),
            2
        )

    return output
```

```python
img = cv2.imread("front.jpg")

img_h, img_w = img.shape[:2]
detector.setInputSize((img_w, img_h))
_, faces = detector.detect(img)
result = draw_yunet_detections(img, faces)
result_rgb = cv2.cvtColor(result, cv2.COLOR_BGR2RGB)
plt.figure(figsize=(10, 7))
plt.imshow(result_rgb)
plt.title("YuNet Face Detection")
plt.axis("off")
plt.show()
```

```python
face_images = [
    "front.jpg",
    "sideface.jpeg",
    "sunglass.jpg",
    "people.jpg"
]

plt.figure(figsize=(16, 12))

for i, image_path in enumerate(face_images):

    img = cv2.imread(image_path)
    img_h, img_w = img.shape[:2]

    detector.setInputSize((img_w, img_h))
    _, faces = detector.detect(img)
    result = draw_yunet_detections(img, faces)

    result_rgb = cv2.cvtColor(
        result,
        cv2.COLOR_BGR2RGB
    )

    plt.subplot(2, 2, i + 1)
    plt.imshow(result_rgb)
    plt.title(
        f"{image_path} | Faces: "
        f"{0 if faces is None else len(faces)}"
    )
    plt.axis("off")

plt.tight_layout()
plt.show()
```

3.3
Real-time detection on webcam

```python
import time
import cv2

cap = cv2.VideoCapture(0)
previous_size = None

frame_count = 0
start_time = time.time()

while True:

    ret, frame = cap.read()

    if not ret:
        print("Could not read frame.")
        break

    h, w = frame.shape[:2]
    current_size = (w, h)

    if current_size != previous_size:
        detector.setInputSize(current_size)
        previous_size = current_size

    _, faces = detector.detect(frame)
    frame = draw_yunet_detections(frame, faces)
    frame_count += 1
    elapsed = time.time() - start_time

    if elapsed > 0:
        fps = frame_count / elapsed
    else:
        fps = 0

    cv2.putText(
        frame,
        f"FPS: {fps:.2f}",
        (20, 40),
        cv2.FONT_HERSHEY_SIMPLEX,
        1,
        (0, 255, 255),
        2
    )

    cv2.imshow("YuNet Real-Time Face Detection", frame)
    key = cv2.waitKey(1) & 0xFF

    if key == 27:
        break

cap.release()
cv2.destroyAllWindows()

print('Q or Esc for exit')
print(f"Average FPS: {fps:.2f}")
```

Score-threshold experiment & applied privacy use-case

```python
def detect_with_threshold(image, threshold):

    h, w = image.shape[:2]
    temp_detector = cv2.FaceDetectorYN.create(
        model_path,
        "",
        (w, h),
        threshold
    )

    temp_detector.setInputSize((w, h))
    _, faces = temp_detector.detect(image)

    return faces
```

```python
test_img = cv2.imread("people.jpg")

thresholds = [0.5, 0.7, 0.9]
plt.figure(figsize=(18, 6))

for i, threshold in enumerate(thresholds):

    faces = detect_with_threshold(test_img,threshold)
    result = draw_yunet_detections(test_img,faces)
    result_rgb = cv2.cvtColor(result,cv2.COLOR_BGR2RGB)

    count = 0 if faces is None else len(faces)

    plt.subplot(1, 3, i + 1)
    plt.imshow(result_rgb)
    plt.title(
        f"Threshold = {threshold}\n"
        f"Faces = {count}"
    )
    plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
def blur_detected_faces(image, faces):
    output = image.copy()

    for face in faces:

        x, y, w, h = face[:4].astype(int)

        x1 = max(0, x)
        y1 = max(0, y)
        x2 = min(output.shape[1], x + w)
        y2 = min(output.shape[0], y + h)

        face_roi = output[y1:y2, x1:x2]

        if face_roi.size == 0:
            continue

        blurred_roi = cv2.GaussianBlur(
            face_roi,
            (51, 51),
            0
        )

        output[y1:y2, x1:x2] = blurred_roi
    return output
```

```python
privacy_img = cv2.imread("people.jpg")

faces = detect_with_threshold(
    privacy_img,
    0.9
)

blurred_img = blur_detected_faces(
    privacy_img,
    faces
)

privacy_rgb = cv2.cvtColor(
    privacy_img,
    cv2.COLOR_BGR2RGB
)

blurred_rgb = cv2.cvtColor(
    blurred_img,
    cv2.COLOR_BGR2RGB
)

plt.figure(figsize=(14, 6))
plt.subplot(1, 2, 1)
plt.imshow(privacy_rgb)
plt.title("Before - Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(blurred_rgb)
plt.title("After - Privacy Blur")
plt.axis("off")

plt.tight_layout()
plt.show()
```

WHY YuNet — Markdown

## WHY YuNet?

YuNet is a compact, modern CNN-based face detector available through OpenCV's FaceDetectorYN interface. It is designed for efficient face detection and is suitable for real-time computer-vision applications.

YuNet returns a face bounding box, five facial landmark points and a confidence score in a single forward detection operation. The five landmarks represent important facial locations such as the two eyes, nose and two mouth corners.

### Why is the score threshold important?

The score threshold determines how confident the detector must be before a detected face is accepted.

A lower threshold such as 0.5 can accept more low-confidence detections, which may increase the number of detected faces but can also allow more false positives.

A higher threshold such as 0.9 is more selective and can reduce weak detections, but it may miss faces that are partially occluded, in unusual poses or poorly illuminated.

Therefore, the threshold represents a trade-off between missed detections and false positives.

### Real-world application

For a real-time security camera, a moderate threshold can be useful when detecting faces under different lighting and poses. For a privacy or photo-tagging application, the threshold should be selected according to the required balance between detecting difficult faces and avoiding incorrect detections.

### Privacy application

After detecting a face, Gaussian blur can be applied only to the detected bounding-box region. This can be used to anonymise people in surveillance footage or images before the content is shared.

```python
def draw_yolo_red(image, results):

    output = image.copy()
    boxes = results[0].boxes

    if boxes is None:
        return output

    for box in boxes:

        x1, y1, x2, y2 = (box.xyxy[0].cpu().numpy().astype(int))

        confidence = float(box.conf[0].cpu().numpy())
        class_id = int(box.cls[0].cpu().numpy())
        class_name = model.names[class_id]

        cv2.rectangle(output,(x1, y1),(x2, y2),(0, 0, 255),2)
        label = f"{class_name} {confidence:.2f}"

        cv2.putText(output,label,(x1, max(y1 - 10, 20)),cv2.FONT_HERSHEY_SIMPLEX,0.6,(0, 0, 255),2)

    return output
```

Morphological pre-cleaning

```python
def morphological_opening(frame):

    gray = cv2.cvtColor(
        frame,
        cv2.COLOR_BGR2GRAY
    )

    _, binary = cv2.threshold(
        gray,
        127,
        255,
        cv2.THRESH_BINARY
    )

    kernel = np.ones(
        (3, 3),
        np.uint8
    )

    opened = cv2.morphologyEx(
        binary,
        cv2.MORPH_OPEN,
        kernel
    )

    return opened
```

```python
def benchmark_yunet(num_frames=100):

    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        raise RuntimeError("Could not open webcam.")

    frame_count = 0
    start_time = time.time()

    while frame_count < num_frames:

        ret, frame = cap.read()
        if not ret:
            break

        h, w = frame.shape[:2]
        detector.setInputSize(
            (w, h)
        )
        detector.detect(frame)

        frame_count += 1

    elapsed = time.time() - start_time
    cap.release()
    fps = frame_count / elapsed
    return fps

yunet_fps = benchmark_yunet(100)
print(f"YuNet Average FPS: {yunet_fps:.2f}")
```

```python
def benchmark_yolo(num_frames=100):

    cap = cv2.VideoCapture(0)

    if not cap.isOpened():
        raise RuntimeError("Could not open webcam.")

    frame_count = 0
    start_time = time.time()

    while frame_count < num_frames:

        ret, frame = cap.read()
        if not ret:
            break

        model(frame,verbose=False)
        frame_count += 1

    elapsed = time.time() - start_time
    cap.release()
    fps = frame_count / elapsed
    return fps

yolo_fps = benchmark_yolo(100)
print(f"YOLO Average FPS: {yolo_fps:.2f}")
```

```python
def benchmark_combined(num_frames=100):

    cap = cv2.VideoCapture(0)
    if not cap.isOpened():
        raise RuntimeError("Could not open webcam.")

    frame_count = 0
    start_time = time.time()
    previous_size = None

    while frame_count < num_frames:

        ret, frame = cap.read()

        if not ret:
            break

        h, w = frame.shape[:2]
        current_size = (w, h)
        if current_size != previous_size:

            detector.setInputSize(current_size)
            previous_size = current_size

        detector.detect(frame)
        model(frame,verbose=False)
        frame_count += 1

    elapsed = time.time() - start_time
    cap.release()
    fps = frame_count / elapsed
    return fps
```

```python
combined_fps = benchmark_combined(100)
print(f"Combined Pipeline FPS: {combined_fps:.2f}")
```

Reflection — Markdown

## Reflection — Integrated Computer Vision Pipeline

The integrated pipeline combines classical image processing with deep-learning-based detection. Classical operations such as morphological opening can be useful for removing small noise or cleaning binary masks, while YuNet and YOLO provide learned detection capabilities for faces and general objects.

### Accuracy vs Compute Trade-off

The choice of model depends on the application. A lightweight model such as YOLOv8n requires fewer computational resources and is suitable for real-time applications on systems with limited hardware. Larger YOLOv8 variants provide greater model capacity but require more computational resources.

The measured FPS values from the 100-frame benchmark show the computational cost of running each detector separately and together. Running YuNet and YOLO simultaneously naturally increases the computational workload compared with running only one detector.

### Real-Time Monitoring

For a real-time monitoring system, the detector must provide sufficient detection quality while maintaining an acceptable frame rate. A slightly lower detection accuracy may sometimes be acceptable if maintaining real-time performance is more important for the application.

### Edge Device vs Cloud

For a low-power edge device, a lightweight model such as YOLOv8n can reduce computational and latency requirements. YuNet's score threshold can also be adjusted according to the required balance between missed detections and false positives.

For a more powerful cloud or server environment, larger YOLOv8 variants can be considered when additional computational resources are available and the application benefits from increased model capacity.

### Overall Pipeline

The experiments demonstrate that classical computer vision and deep learning can complement each other. Morphological operations and histograms are useful for image analysis and preprocessing, while YuNet is specialized for face detection and YOLO is designed for general object detection. Combining these techniques allows a single computer-vision system to perform different types of analysis on the same camera stream.
